```python
import pandas as pd
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis
```

```python
# Caminho do arquivo FASTA
arquivo_fasta = "/content/amps.fasta"

def caracterizar_amps(arquivo_fasta, caminho_csv_saida):
    """
    Lê um arquivo FASTA contendo AMPs vegetais,
    calcula propriedades físico-químicas e exporta os
    resultados para um arquivo CSV.
    """
    resultados = [] # Lista que irá armazenar os dados de cada sequência

    for record in SeqIO.parse(arquivo_fasta, "fasta"):

        seq_str = str(record.seq)
        protein = ProteinAnalysis(seq_str)  # Objeto usado para calcular as propriedades
        massa_kda = protein.molecular_weight() / 1000

        resultados.append({
            "ID": record.id,
            "Comprimento (aa)": len(seq_str),
            "Massa Molecular (kDa)": round(massa_kda, 3),
            "pI": round(protein.isoelectric_point(), 3),
            "Índice de Instabilidade": round(protein.instability_index(), 3),
            "Aromaticidade": round(protein.aromaticity(), 4)
        })

    # Organiza todos os resultados em uma tabela
    df = pd.DataFrame(resultados)

    # Salva a tabela em um arquivo CSV
    df.to_csv(caminho_csv_saida, index=False)
    return df

saida = "/content/caracterizacao_amps.csv"

# Executa a função e armazena o resultado em df
df = caracterizar_amps(arquivo_fasta, saida)

df.head()
```
